```python
from pathlib import Path

from car_eval_io import app
from car_eval_io.config import load_config
from car_eval_io.paths import RunPaths
from car_eval_io.sources import LocalFile, UciDownload  # noqa: F401  (UciDownload() downloads instead)

PROJECT = Path("/content/drive/MyDrive/AME_505/Final_Project")

config = load_config(Path("/content/car-evaluation/configs/config.yaml"), profile="colab")
outcome = app.run(config, data=LocalFile(PROJECT / "car.data"), paths=RunPaths(PROJECT / "outputs"))
```
